```python
import pandas as pd

path_to_data = "../data/matchings_2_supervisors.csv"
matching_df = pd.read_csv(path_to_data)

# Get the index of rows where "id_scopus_student" is -1
mask = matching_df["id_scopus_student"] == -1

# Use .loc to modify the values in-place
matching_df.loc[mask, "id_scopus_student"] = -1 * matching_df.loc[mask].index

matching_df.to_csv(path_to_data)
```
