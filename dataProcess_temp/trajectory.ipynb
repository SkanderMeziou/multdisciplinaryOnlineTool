```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from collections import defaultdict
from unidecode import unidecode
from sklearn.manifold import TSNE

# --- PARAMÈTRES À ADAPTER ---
input_file = "../data/author_publications_dir/A.parquet"      # Fichier Parquet des chercheurs (par lettre)
coord_csv = "../data/coordinates.csv" # Ton CSV disciplines/coords
output_html = "traj_moyenne_X.html"

# --- 1. Charger le CSV des disciplines ---
df_coord = pd.read_csv(coord_csv, encoding="utf-8")
disciplines = df_coord.iloc[:, 0].tolist()          # Noms des disciplines (première colonne)
matrix_coord = df_coord.iloc[:, 1:].to_numpy()      # Le reste (features)

# --- 2. Calculer l'embedding TSNE ---
embedded = TSNE(n_components=2, learning_rate='auto', random_state=42, perplexity=5).fit_transform(matrix_coord)
# Création du mapping discipline → coordonnées 2D
disc_to_coord = {disc: emb for disc, emb in zip(disciplines, embedded)}

# --- 3. Fonction pour obtenir la trajectoire annuelle d'un auteur ---
def get_traj(pub_df, author, disc_to_coord):
    norm = lambda s: unidecode(s.strip().lower())
    author_norm = norm(author)
    pub_df["author_norm"] = pub_df["author"].apply(norm)
    df_filtered = pub_df[pub_df["author_norm"] == author_norm]
    pub_dict = defaultdict(lambda: defaultdict(int))
    for _, row in df_filtered.iterrows():
        discipline = row["discipline"]
        year = row["year"]
        if pd.notnull(year) and pd.notnull(discipline):
            pub_dict[int(year)][discipline] += 1
    # Trajectoire : barycentre pour chaque année
    traj = []
    years = sorted(pub_dict)
    for year in years:
        dico = pub_dict[year]
        total = sum(dico.values())
        coords = np.array([
            disc_to_coord[disc] * n / total
            for disc, n in dico.items()
            if disc in disc_to_coord
        ])
        if len(coords) == 0:
            continue
        bary = np.sum(coords, axis=0)
        traj.append(bary)
    return np.array(traj)

# --- 4. Charger les auteurs du fichier Parquet ---
df_letter = pd.read_parquet(input_file)
col_author = "author" if "author" in df_letter.columns else "author_name"
authors = df_letter[col_author].drop_duplicates().tolist()

# --- 5. Calcul trajectoires individuelles ---
all_dirs = []
all_dists = []
for author in authors:
    traj = get_traj(df_letter, author, disc_to_coord)
    if len(traj) < 2:
        continue
    dirs = traj[1:] - traj[:-1]
    norms = np.linalg.norm(dirs, axis=1, keepdims=True)
    dirs_normed = dirs / (norms + 1e-9)
    all_dirs.append(dirs_normed)
    total_dist = np.sum(norms)
    all_dists.append(total_dist)

if not all_dirs:
    print("Aucune trajectoire exploitable trouvée.")
    exit(1)

# --- 6. Moyenne par étape ---
max_len = max(d.shape[0] for d in all_dirs)
dir_mean = []
for i in range(max_len):
    valid = [d[i] for d in all_dirs if d.shape[0] > i]
    if valid:
        dir_mean.append(np.mean(valid, axis=0))
    else:
        dir_mean.append(np.array([np.nan, np.nan]))
dir_mean = np.array(dir_mean)
final_distance_mean = np.mean(all_dists)

# --- 7. Trajectoire moyenne cumulative ---
traj_mean = [np.array([0.0, 0.0])]
for d in dir_mean:
    traj_mean.append(traj_mean[-1] + d)
traj_mean = np.array(traj_mean)

# --- 8. Visualisation Plotly ---
fig = go.Figure()

# Origine
fig.add_trace(go.Scatter(
    x=[traj_mean[0,0]],
    y=[traj_mean[0,1]],
    mode='markers+text',
    marker=dict(size=14, color='green'),
    text=["Départ"],
    textposition="top center",
    name="Départ"
))

# Arrivée
fig.add_trace(go.Scatter(
    x=[traj_mean[-1,0]],
    y=[traj_mean[-1,1]],
    mode='markers+text',
    marker=dict(size=14, color='red'),
    text=["Arrivée"],
    textposition="bottom right",
    name="Arrivée"
))

# Segments/flèches de la trajectoire moyenne
for i in range(len(traj_mean)-1):
    fig.add_trace(go.Scatter(
        x=[traj_mean[i,0], traj_mean[i+1,0]],
        y=[traj_mean[i,1], traj_mean[i+1,1]],
        mode="lines+markers+text",
        line=dict(width=4, color="blue"),
        marker=dict(size=10, color="blue"),
        text=[None, f"Étape {i+1}"],
        showlegend=False
    ))
    fig.add_annotation(
        ax=traj_mean[i,0],
        ay=traj_mean[i,1],
        x=traj_mean[i+1,0],
        y=traj_mean[i+1,1],
        showarrow=True,
        arrowhead=3,
        arrowsize=1,
        arrowwidth=2,
        arrowcolor="blue"
    )

fig.update_layout(
    title=f"Trajectoire moyenne (chercheurs {input_file})<br>Distance moyenne parcourue : {final_distance_mean:.2f}",
    xaxis=dict(showgrid=True, zeroline=True),
    yaxis=dict(showgrid=True, zeroline=True),
    width=800,
    height=600
)

fig.write_html(output_html)
print(f"✅ Résultat enregistré dans : {output_html}")
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from collections import defaultdict
from pathlib import Path
from unidecode import unidecode

# === 1. Chargement des coordonnées t-SNE ===
df_coords = pd.read_csv("../data/coordinates.csv")
disciplines = df_coords["disc"].tolist()
matrix = df_coords.drop(columns=["disc"]).to_numpy()

# Vérifie automatiquement les noms de colonnes coordonnées
coord_cols = [c for c in df_coords.columns if c != "disc"]
disc_to_coord = {
    row["disc"]: row[coord_cols].to_numpy()
    for _, row in df_coords.iterrows()
}

# === 2. Chargement des publications auteur ===
df_pub = pd.read_parquet("../data/author_publications.parquet")
def plot_author_trajectory_fast(author_name, window_size=None):
    pub_data = get_author_publications_fast(author_name)
    if not pub_data:
        print(f"Aucune publication trouvée pour {author_name}")
        return

    trajectory = []
    all_years = sorted(pub_data)

    for idx, year in enumerate(all_years):
        if window_size is None:
            # Mode cumulatif
            selected_years = all_years[:idx + 1]
        else:
            # Mode fenêtre glissante
            selected_years = [
                y for y in all_years
                if year - window_size < y <= year
            ]
        if not selected_years:
            continue

        aggregated = defaultdict(int)
        for y in selected_years:
            for disc, count in pub_data[y].items():
                aggregated[disc] += count

        total = sum(aggregated.values())
        coords = np.array([
            disc_to_coord[disc] * weight / total
            for disc, weight in aggregated.items()
            if disc in disc_to_coord
        ])
        if len(coords) == 0:
            continue
        bary = np.sum(coords, axis=0)
        trajectory.append((year, bary))

    # === Tracé Plotly ===
    fig = go.Figure()

    # 1. Disciplines en fond
    for disc, coord in disc_to_coord.items():
        fig.add_trace(go.Scatter(
            x=[coord[0]],
            y=[coord[1]],
            mode='markers+text',
            marker=dict(size=8, color='lightgray'),
            text=[disc],
            textposition='top center',
            showlegend=False
        ))

    # 2. Trajectoire
    x_vals = [pt[1][0] for pt in trajectory]
    y_vals = [pt[1][1] for pt in trajectory]
    years  = [pt[0] for pt in trajectory]

    fig.add_trace(go.Scatter(
        x=x_vals,
        y=y_vals,
        mode='lines+markers+text',
        line=dict(color='blue', width=3),
        marker=dict(size=10, color='blue'),
        text=[str(y) for y in years],
        textposition='bottom right',
        name=f"Évolution de {author_name.title()}"
    ))

    # 3. Flèches
    for i in range(len(trajectory) - 1):
        fig.add_annotation(
            ax=trajectory[i][1][0],
            ay=trajectory[i][1][1],
            x=trajectory[i+1][1][0],
            y=trajectory[i+1][1][1],
            showarrow=True,
            arrowhead=30,
            arrowsize=1,
            arrowwidth=2,
            arrowcolor="blue"
        )

    mode_label = f" (fenêtre {window_size} ans)" if window_size else " (cumulatif)"
    fig.update_layout(
        title=f"Trajectoire disciplinaire de {author_name.title()}{mode_label}",
        xaxis=dict(showgrid=False, zeroline=False, visible=False),
        yaxis=dict(showgrid=False, zeroline=False, visible=False),
        legend=dict(x=0.02, y=0.98)
    )

    fig.show()
    fig.write_html(f"trajectory_{author_name.replace(' ', '_')}_{'sliding' if window_size else 'cumulative'}.html")

plot_author_trajectory_fast("Elie R.",5)                # cumulatif
 # fenêtre 5 ans
```

```python
import pandas as pd
import numpy as np
import plotly.express as px
from collections import defaultdict

# === 1. Chargement des coordonnées t-SNE ===
df_coords = pd.read_csv("../data/coordinates.csv")
coord_cols = [c for c in df_coords.columns if c != "disc"]
disc_to_coord = {
    row["disc"]: row[coord_cols].to_numpy()
    for _, row in df_coords.iterrows()
}

# === 2. Chargement des publications auteur ===
df_pub = pd.read_parquet("../data/author_publications_dir/X.parquet")
# Colonnes : ['author', 'discipline', 'year']

# === 3. Fonction pour récupérer publications d'un auteur ===
def get_author_publications_fast(author_name):
    sub = df_pub[df_pub["author"].str.lower() == author_name.lower()]
    if sub.empty:
        return {}

    # Compte le nombre de publications par discipline et par année
    grouped = sub.groupby(["year", "discipline"]).size().reset_index(name="count")

    result = defaultdict(lambda: defaultdict(int))
    for _, row in grouped.iterrows():
        result[row["year"]][row["discipline"]] = row["count"]
    return result

# === 4. Calcul distance début-fin pour un auteur ===
def compute_start_end_distance(author_name):
    pub_data = get_author_publications_fast(author_name)
    if not pub_data:
        return 0, 0  # (nb_publications, distance_debut_fin)

    all_years = sorted(pub_data)
    total_pubs = sum(sum(pub_data[y].values()) for y in pub_data)

    barycenters = []
    for year in all_years:
        aggregated = defaultdict(int)
        # cumul jusqu'à cette année
        for y in all_years[:all_years.index(year) + 1]:
            for disc, count in pub_data[y].items():
                aggregated[disc] += count

        total = sum(aggregated.values())
        coords = np.array([
            disc_to_coord[disc] * weight / total
            for disc, weight in aggregated.items()
            if disc in disc_to_coord
        ])
        if len(coords) == 0:
            continue
        barycenters.append(np.sum(coords, axis=0))

    if len(barycenters) < 2:
        return total_pubs, 0  # pas de mouvement mesurable

    # Distance entre le 1er et le dernier barycentre
    start_end_distance = np.linalg.norm(barycenters[-1] - barycenters[0])
    return total_pubs, start_end_distance

# === 5. Prendre un échantillon de 10000 auteurs aléatoires ===
authors = df_pub["author"].unique()
sample_authors = np.random.choice(authors, size=min(10000, len(authors)), replace=False)

# === 6. Calcul pour l'échantillon ===
records = []
for author in sample_authors:
    total_pubs, start_end_dist = compute_start_end_distance(author)
    records.append({"author": author, "publications": total_pubs, "distance": start_end_dist})

df_analysis = pd.DataFrame(records)

# === 7. Scatter plot ===
fig = px.scatter(
    df_analysis,
    x="publications",
    log_x=True,
    y="distance",
    hover_name="author",
    title="Corrélation entre nombre de publications et distance début-fin (échantillon 10000)",
    labels={"publications": "Nombre de publications", "distance": "Distance début-fin"}
)
fig.show()

# === 8. Corrélation Pearson ===
corr = df_analysis[["publications", "distance"]].corr().iloc[0, 1]
print(f"Corrélation (échantillon) : {corr:.3f}")
```

```python
# -*- coding: utf-8 -*-
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from collections import defaultdict

# === t-SNE & prétraitement ===
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# ------------------------------------------------------------------
# 1) Charger df_coords (features par discipline) et calculer un t-SNE 2D
#    Hypothèse : df_coords a une colonne 'disc' + colonnes features numériques
df_coords_raw = pd.read_csv("../data/coordinates.csv")

assert "disc" in df_coords_raw.columns, "Il faut une colonne 'disc' dans coordinates.csv"
feature_cols = [c for c in df_coords_raw.columns if c != "disc"]

# Nettoyage basique
X = df_coords_raw[feature_cols].replace([np.inf, -np.inf], np.nan).fillna(0.0).to_numpy(dtype=np.float32)

# Standardisation + PCA (accélère t-SNE si beaucoup de dimensions)
scaler = StandardScaler(with_mean=True, with_std=True)
X_std = scaler.fit_transform(X)

# Garde jusqu'à 50 composantes max pour stabiliser/accélérer
n_comp = min(50, X_std.shape[1])
X_pca = PCA(n_components=n_comp, random_state=42).fit_transform(X_std)

# Perplexité adaptée (≃ nb voisins), borne par (n-1)/3
n_samples = X_pca.shape[0]
default_perp = 30
perp = max(5, min(default_perp, (n_samples - 1) / 3))

tsne = TSNE(
    n_components=2,
    perplexity=perp,
    learning_rate="auto",
    init="pca",
    metric="euclidean",
    n_iter=1000,
    random_state=42,
    verbose=0
)
tsne_2d = tsne.fit_transform(X_pca)  # shape: (n_disc, 2)

# Construire le DataFrame t-SNE et le mapping
df_tsne = pd.DataFrame({
    "disc": df_coords_raw["disc"].values,
    "TSNE-1": tsne_2d[:, 0],
    "TSNE-2": tsne_2d[:, 1],
})
disc_to_coord = {
    row["disc"]: np.array([row["TSNE-1"], row["TSNE-2"]], dtype=np.float32)
    for _, row in df_tsne.iterrows()
}

# (Optionnel) Sauvegarde pour réutilisation
# df_tsne.to_csv("../data/coordinates_tsne2d.csv", index=False)

# ------------------------------------------------------------------
# 2) Charger les publications
df_pub = pd.read_parquet("../data/author_publications_dir/U.parquet")
# Colonnes attendues : ['author', 'discipline', 'year']

# Fonctions utilitaires
def get_author_publications_fast(author_name: str):
    sub = df_pub[df_pub["author"].str.lower() == author_name.lower()]
    if sub.empty:
        return {}
    grouped = sub.groupby(["year", "discipline"]).size().reset_index(name="count")
    d = defaultdict(lambda: defaultdict(int))
    for _, r in grouped.iterrows():
        d[int(r["year"])][r["discipline"]] = int(r["count"])
    return d
def author_trajectory(author_name: str):
    """
    Trajectoire « instantanée » 2D d'un·e auteur·e.

    Pour chaque année t où l'auteur a publié, on calcule
    un point 2D = barycentre pondéré des disciplines publiées
    *uniquement* durant l'année t (pas de cumul avec les années précédentes).

    Retour :
        traj : list[np.array([x, y])]  # un point par année chronologique
    """
    # 1) Récupérer les publications par discipline et par année
    pub_data = get_author_publications_fast(author_name)
    if not pub_data:
        return []

    # 2) Trier les années
    all_years = sorted(pub_data)           # ex. [2010, 2011, 2014, …]

    traj = []
    for y in all_years:
        year_disc_counts = pub_data[y]     # {discipline: nb_articles} pour l'année y
        tot = sum(year_disc_counts.values())
        if tot == 0:
            continue

        # 3) Construire le barycentre de L'ANNÉE SEULEMENT
        weighted_coords = []
        for disc, n in year_disc_counts.items():
            if disc in disc_to_coord:      # on ignore les disciplines sans coord.
                weight = n / tot           # proportion d'articles
                weighted_coords.append(disc_to_coord[disc] * weight)

        if not weighted_coords:            # aucune discipline connue => on saute
            continue

        coords = np.sum(weighted_coords, axis=0)   # barycentre 2D pour l'année y
        traj.append(coords)

    return traj   # ex. [array([x0, y0]), array([x1, y1]), …]

# ------------------------------------------------------------------
# 3) Même échantillon que d'habitude
authors = df_pub["author"].unique()
sample_authors = np.random.choice(authors, size=min(5000, len(authors)), replace=False)

# 4) Collecte des déplacements consécutifs p(t) -> p(t+1)
starts, vecs = [], []
for a in sample_authors:
    traj = author_trajectory(a)
    for i in range(len(traj) - 1):
        p0, p1 = traj[i], traj[i + 1]
        starts.append(p0)
        vecs.append(p1 - p0)

if len(starts) == 0:
    raise RuntimeError("Aucune trajectoire utilisable pour construire le champ de vecteurs.")

starts = np.vstack(starts)[:, :2]  # en 2D
vecs   = np.vstack(vecs)[:, :2]

# ------------------------------------------------------------------
# 5) Quadrillage + agrégation des vecteurs par cellule de départ
# Taille de cellule (en unités t-SNE) : ajuste selon la densité
cell_size = 2.0

xmin, ymin = starts.min(axis=0).tolist()
xmax, ymax = starts.max(axis=0).tolist()

ix = np.floor((starts[:, 0] - xmin) / cell_size).astype(int)
iy = np.floor((starts[:, 1] - ymin) / cell_size).astype(int)

grid = defaultdict(list)
for i in range(len(ix)):
    grid[(ix[i], iy[i])].append(vecs[i])

centers, finals, counts = [], [], []
for (gx, gy), vlist in grid.items():
    v_arr = np.vstack(vlist)
    mean_vec = v_arr.mean(axis=0)            # vecteur moyen (dx, dy)
    cx = xmin + (gx + 0.5) * cell_size       # centre de cellule
    cy = ymin + (gy + 0.5) * cell_size
    centers.append([cx, cy])
    finals.append(mean_vec)
    counts.append(len(vlist))                # nb de transitions agrégées

centers = np.array(centers)
finals  = np.array(finals)
counts  = np.array(counts)

# ------------------------------------------------------------------
# 6) Visualisation du champ de vecteurs
scale = 0.5 # allonge/raccourcit les flèches

x0, y0 = centers[:, 0], centers[:, 1]
x1, y1 = centers[:, 0] + finals[:, 0] * scale, centers[:, 1] + finals[:, 1] * scale

# Couleur par densité (nb. de transitions par cellule) – optionnel
# On normalise pour un rendu simple
if counts.size > 0:
    cmin, cmax = counts.min(), counts.max()
    if cmax == cmin:
        colors = ["rgba(200,0,0,0.7)"] * len(counts)
    else:
        # plus il y a de vecteurs agrégés, plus la flèche est opaque
        alphas = 0.3 + 0.7 * (counts - cmin) / (cmax - cmin)
        colors = [f"rgba(200,0,0,{alpha:.3f})" for alpha in alphas]
else:
    colors = ["rgba(200,0,0,0.7)"] * len(x0)

# … (tout le code précédent inchangé jusqu’à la construction de fig) …

import plotly.express as px

# ------------------------------------------------------------------
# Palette de couleurs : assez longue pour couvrir beaucoup de disciplines
import plotly.express as px
from plotly.colors import sample_colorscale

# ------------------------------------------------------------------
# Palette Turbo (disponible dans px.colors.sequential)
# On prélève autant de couleurs qu'il y a de disciplines
n_disc = df_tsne["disc"].nunique()
palette = sample_colorscale(
    px.colors.sequential.Turbo,
    [i / (n_disc - 1) for i in range(n_disc)]
)
disc_to_color = {d: palette[i] for i, d in enumerate(df_tsne["disc"].unique())}

# ------------------------------------------------------------------
fig = go.Figure()

# 1) Points + libellés des disciplines -----------------------------
fig.add_trace(
    go.Scatter(
        x=df_tsne["TSNE-1"],
        y=df_tsne["TSNE-2"],
        mode="markers+text",
        marker=dict(
            size=6,
            color=[disc_to_color[d] for d in df_tsne["disc"]],
            opacity=0.6
        ),
        text=df_tsne["disc"],
        textfont=dict(size=9, color="rgba(50,50,50,0.7)"),
        textposition="top center",
        hovertemplate="<b>%{text}</b><extra></extra>",
        name="Disciplines"
    )
)

# 2) Champ de vecteurs (flèches) — on ne trace que si >= 11 transitions
for i in range(len(x0)):
    if counts[i] <= 10:
        continue                       # on saute les cellules « peu peuplées »

    fig.add_trace(go.Scatter(
        x=[x0[i], x1[i]], y=[y0[i], y1[i]],
        mode="lines",
        line=dict(color=colors[i], width=2),
        hoverinfo="text",
        text=f"Transitions agrégées : {counts[i]}",
        showlegend=False
    ))
    fig.add_trace(go.Scatter(
        x=[x0[i]], y=[y0[i]],
        mode="markers",
        marker=dict(size=3, color=colors[i]),
        hoverinfo="skip",
        showlegend=False
    ))
# 2) Champ de vecteurs : vraies flèches (annotations Plotly)
for i in range(len(x0)):
    if counts[i] <= 10:
        continue  # on n'affiche pas les cellules < 11 transitions

    fig.add_annotation(
        x=x1[i], y=y1[i],            # extrémité de la flèche
        ax=x0[i], ay=y0[i],          # origine
        xref="x", yref="y", axref="x", ayref="y",
        showarrow=True,
        arrowhead=1,               
        arrowsize=1.2,            
        arrowwidth=1,
        arrowcolor=colors[i],
        opacity=1.0,
        hovertext=f"Transitions agrégées : {counts[i]}",
        hoverlabel=dict(bgcolor="rgba(255,255,255,0.8)")
    )


fig.update_layout(
    title="Champ de vecteurs + positions & libellés des disciplines (t‑SNE 2D)",
    xaxis=dict(scaleanchor="y", scaleratio=1),
    width=950, height=820
)

fig.show()
fig.write_html("champVect.html")
```

```python
# -*- coding: utf-8 -*-
import pandas as pd
import numpy as np
import plotly.express as px
from collections import defaultdict
import random

# ===============================
# 0) Paramètres
# ===============================
PATH_COORDS = "../data/coordinates.csv"  # colonnes: disc, dim1, dim2, ...
PATH_PUBS   = "../data/author_publications_dir/X.parquet"  # colonnes: author, discipline, year
MIN_PUBS_FILTER = 50         # seuil d'éligibilité des auteurs
SAMPLE_MAX = 5000            # échantillon d'auteurs max
BINS = 50                    # nb de bacs pour les histogrammes
SEED = 42                    # reproductibilité

random.seed(SEED)
np.random.seed(SEED)

# ===============================
# 1) Chargement des coordonnées t-SNE/UMAP
# ===============================
df_coords = pd.read_csv(PATH_COORDS)
coord_cols = [c for c in df_coords.columns if c != "disc"]
disc_to_coord = {
    row["disc"]: row[coord_cols].to_numpy(dtype=np.float32)
    for _, row in df_coords.iterrows()
}

# ===============================
# 2) Chargement des publications auteur
# ===============================
df_pub = pd.read_parquet(PATH_PUBS)
# Colonnes attendues : ['author', 'discipline', 'year']

# ===============================
# 3) Récup publications d'un auteur (rapide)
# ===============================
def get_author_publications_fast(author_name: str):
    sub = df_pub[df_pub["author"].str.lower() == author_name.lower()]
    if sub.empty:
        return None

    grouped = sub.groupby(["year", "discipline"]).size().reset_index(name="count")
    pub_data = defaultdict(lambda: defaultdict(int))
    for _, row in grouped.iterrows():
        pub_data[int(row["year"])][row["discipline"]] = int(row["count"])
    return pub_data

# ===============================
# 4) Vitesse moyenne et longueur de trajectoire
# ===============================
def compute_average_speed(author_name: str):
    """
    Retourne:
      (total_pubs, total_distance, avg_speed)
    - total_distance = somme des distances inter-annuelles de la trajectoire cumulée
    - avg_speed = total_distance / nb_annees_avec_publications
    """
    pub_data = get_author_publications_fast(author_name)
    if not pub_data:
        return 0, 0.0, 0.0

    all_years = sorted(pub_data)
    total_pubs = sum(sum(pub_data[y].values()) for y in pub_data)
    if total_pubs <= 20:
        return total_pubs, 0.0, 0.0

    trajectory = []
    aggregated = defaultdict(int)

    for year in all_years:
        # cumule les publications jusqu'à l'année courante
        for disc, count in pub_data[year].items():
            aggregated[disc] += count

        total = sum(aggregated.values())
        if total == 0:
            continue

        # barycentre pondéré des disciplines connues
        weighted = []
        for d, w in aggregated.items():
            if d in disc_to_coord:
                weighted.append(disc_to_coord[d] * w)
        if not weighted:
            continue

        coords = np.sum(weighted, axis=0) / total
        trajectory.append(coords)

    if len(trajectory) < 2:
        return total_pubs, 0.0, 0.0

    trajectory = np.vstack(trajectory)
    diffs = trajectory[1:] - trajectory[:-1]
    total_distance = float(np.sum(np.linalg.norm(diffs, axis=1)))

    nb_years = len(all_years)  # années avec au moins 1 publi (après groupby)
    avg_speed = total_distance / nb_years if nb_years > 0 else 0.0

    return total_pubs, total_distance, avg_speed

# ===============================
# 5) Auteurs éligibles
# ===============================
pub_counts = df_pub.groupby("author").size()
eligible_authors = pub_counts[pub_counts > MIN_PUBS_FILTER].index.to_numpy()

if len(eligible_authors) > SAMPLE_MAX:
    sample_authors = np.random.choice(eligible_authors, size=SAMPLE_MAX, replace=False)
else:
    sample_authors = eligible_authors

# ===============================
# 6) Calcul pour l'échantillon
# ===============================
records = []
for author in sample_authors:
    total_pubs, total_dist, avg_speed = compute_average_speed(author)
    if total_pubs > 20:  # garde une petite barrière de bruit
        records.append({
            "author": author,
            "publications": int(total_pubs),
            "distance": float(total_dist),
            "avg_speed": float(avg_speed)
        })

df_analysis = pd.DataFrame(records)

# Sanity-check rapide
if df_analysis.empty:
    raise RuntimeError("df_analysis est vide. Vérifie les chemins, colonnes et filtres (MIN_PUBS_FILTER).")

# ===============================
# 7) (Optionnel) Scatter publications vs vitesse
# ===============================
fig_scatter = px.scatter(
    df_analysis,
    x="publications",
    y="avg_speed",
    log_x=True,
    hover_name="author",
    title=f"Vitesse disciplinaire moyenne (auteurs avec > {MIN_PUBS_FILTER} publications)",
    labels={"publications": "Nombre de publications", "avg_speed": "Vitesse moyenne (distance / années)"},
)
fig_scatter.show()

corr = df_analysis[["publications", "avg_speed"]].corr().iloc[0, 1]
print(f"Corrélation (auteurs > {MIN_PUBS_FILTER} pubs) : {corr:.3f}")

# ===============================
# 8) Distributions (Plotly)
# ===============================

# --- Histogramme des vitesses moyennes ---
fig_speed = px.histogram(
    df_analysis,
    x="avg_speed",
    nbins=BINS,
    title="Distribution des vitesses moyennes",
    labels={"avg_speed": "Vitesse moyenne (distance / années)"},
    marginal="rug",          # petit rug plot au-dessus
    opacity=0.9
)
fig_speed.update_layout(bargap=0.05)
fig_speed.show()

# --- Histogramme des longueurs disciplinaires (distance totale) ---
fig_dist = px.histogram(
    df_analysis,
    x="distance",
    nbins=BINS,
    title="Distribution des longueurs disciplinaires (distance totale)",
    labels={"distance": "Distance totale (longueur de trajectoire)"},
    marginal="rug",
    opacity=0.9
)
fig_dist.update_layout(bargap=0.05)
fig_dist.show()

# ===============================
# 9) Courbes cumulées (ECDF) utiles pour lire les quantiles
# ===============================
fig_speed_ecdf = px.ecdf(
    df_analysis,
    x="avg_speed",
    title="ECDF des vitesses moyennes",
    labels={"avg_speed": "Vitesse moyenne (distance / années)"}
)
fig_speed_ecdf.show()

fig_dist_ecdf = px.ecdf(
    df_analysis,
    x="distance",
    title="ECDF des longueurs disciplinaires (distance totale)",
    labels={"distance": "Distance totale (longueur de trajectoire)"}
)
fig_dist_ecdf.show()

# ===============================
# 10) (Optionnel) Stats rapides
# ===============================
def quick_stats(series: pd.Series, name: str):
    q = series.quantile([0.05, 0.25, 0.5, 0.75, 0.95])
    print(f"\n{name} — stats rapides")
    print(f"  n = {series.notna().sum()}")
    print(f"  min = {series.min():.4f}")
    print(f"  q05 = {q.loc[0.05]:.4f}")
    print(f"  q25 = {q.loc[0.25]:.4f}")
    print(f"  q50 = {q.loc[0.50]:.4f}")
    print(f"  q75 = {q.loc[0.75]:.4f}")
    print(f"  q95 = {q.loc[0.95]:.4f}")
    print(f"  max = {series.max():.4f}")
    print(f"  mean = {series.mean():.4f}")
    print(f"  std = {series.std():.4f}")

quick_stats(df_analysis["avg_speed"], "Vitesse moyenne")
quick_stats(df_analysis["distance"], "Distance totale")

# === 11) Sauvegarde en HTML ===
fig_scatter.write_html("scatter_publications_vs_speed.html")
fig_speed.write_html("histogram_avg_speed.html")
fig_dist.write_html("histogram_distance.html")
fig_speed_ecdf.write_html("ecdf_avg_speed.html")
fig_dist_ecdf.write_html("ecdf_distance.html")

print("✅ Tous les graphiques ont été sauvegardés en HTML !")
```

```python
# -*- coding: utf-8 -*-
import pandas as pd
import numpy as np
import plotly.express as px
from collections import defaultdict
import random

# ===============================
# 0) Paramètres
# ===============================
PATH_COORDS = "../data/coordinates.csv"             # colonnes: disc, dim1, dim2, ...
PATH_PUBS   = "../data/author_publications_dir/X.parquet"  # colonnes: author, discipline, year
MIN_PUBS_FILTER = 50       # seuil d'éligibilité des auteurs
SAMPLE_MAX = 5000          # échantillon max d'auteurs
BINS = 50                  # nb de bacs pour les histogrammes
SEED = 42                  # reproductibilité

random.seed(SEED)
np.random.seed(SEED)

# ===============================
# 1) Chargement des coordonnées t-SNE/UMAP
# ===============================
df_coords = pd.read_csv(PATH_COORDS)
coord_cols = [c for c in df_coords.columns if c != "disc"]
disc_to_coord = {
    row["disc"]: row[coord_cols].to_numpy(dtype=np.float32)
    for _, row in df_coords.iterrows()
}

# ===============================
# 2) Chargement des publications
# ===============================
df_pub = pd.read_parquet(PATH_PUBS)
# Colonnes attendues : ['author', 'discipline', 'year']

# ===============================
# 3) Vitesse "publication par publication"
# ===============================
def compute_average_speed_per_pub(author_name: str):
    """
    Calcule la trajectoire en ajoutant les publications une par une (ordre chronologique),
    et renvoie (total_pubs, total_distance, avg_speed)
      - total_distance = somme des distances entre barycentres successifs (à chaque nouvelle publi)
      - avg_speed = total_distance / nb_annees_de_publication (normalisation temporelle inchangée)
    """
    sub = df_pub[df_pub["author"].str.lower() == author_name.lower()]
    if sub.empty:
        return 0, 0.0, 0.0

    total_pubs = len(sub)
    if total_pubs <= 20:
        return total_pubs, 0.0, 0.0

    # trie chronologiquement (si égalité intra-année, l'ordre reste celui du DataFrame)
    sub_sorted = sub.sort_values(by=["year"]).reset_index(drop=True)

    trajectory = []
    aggregated = defaultdict(int)

    for _, row in sub_sorted.iterrows():
        disc = row["discipline"]
        aggregated[disc] += 1  # on ajoute UNE publi

        total = sum(aggregated.values())
        if total == 0:
            continue

        # barycentre pondéré sur les disciplines connues
        weighted = []
        for d, w in aggregated.items():
            if d in disc_to_coord:
                weighted.append(disc_to_coord[d] * w)
        if not weighted:
            continue

        coords = np.sum(weighted, axis=0) / total
        trajectory.append(coords)

    if len(trajectory) < 2:
        return total_pubs, 0.0, 0.0

    trajectory = np.vstack(trajectory)
    diffs = trajectory[1:] - trajectory[:-1]
    total_distance = float(np.sum(np.linalg.norm(diffs, axis=1)))

    # nb d'années distinctes avec publications (normalisation identique à avant)
    nb_years = sub["year"].nunique()
    avg_speed = total_distance / nb_years if nb_years > 0 else 0.0

    return total_pubs, total_distance, avg_speed

# ===============================
# 4) Auteurs éligibles & échantillon
# ===============================
pub_counts = df_pub.groupby("author").size()
eligible_authors = pub_counts[pub_counts > MIN_PUBS_FILTER].index.to_numpy()

if len(eligible_authors) > SAMPLE_MAX:
    sample_authors = np.random.choice(eligible_authors, size=SAMPLE_MAX, replace=False)
else:
    sample_authors = eligible_authors

# ===============================
# 5) Calculs (granularité "par publication")
# ===============================
records_pub = []
for author in sample_authors:
    total_pubs, total_dist, avg_speed = compute_average_speed_per_pub(author)
    if total_pubs > 20:
        records_pub.append({
            "author": author,
            "publications": int(total_pubs),
            "distance": float(total_dist),
            "avg_speed": float(avg_speed),
        })

df_analysis_pub = pd.DataFrame(records_pub)

if df_analysis_pub.empty:
    raise RuntimeError("df_analysis_pub est vide. Vérifie chemins/données/filtres.")

# ===============================
# 6) Corrélation & Scatter (par publication)
# ===============================
corr_pub = df_analysis_pub[["publications", "avg_speed"]].corr().iloc[0, 1]
print(f"Corrélation (granularité par publication, auteurs > {MIN_PUBS_FILTER} pubs) : {corr_pub:.3f}")

fig_scatter_pub = px.scatter(
    df_analysis_pub,
    x="publications",
    y="avg_speed",
    log_x=True,
    hover_name="author",
    title="Vitesse disciplinaire moyenne — granularité par publication",
    labels={"publications": "Nombre de publications", "avg_speed": "Vitesse moyenne (distance / années)"},
)
fig_scatter_pub.show()

# ===============================
# 7) Distributions (histogrammes)
# ===============================
fig_speed = px.histogram(
    df_analysis_pub,
    x="avg_speed",
    nbins=BINS,
    title="Distribution des vitesses moyennes (granularité par publication)",
    labels={"avg_speed": "Vitesse moyenne (distance / années)"},
    marginal="rug",
    opacity=0.9,
)
fig_speed.update_layout(bargap=0.05)
fig_speed.show()

fig_dist = px.histogram(
    df_analysis_pub,
    x="distance",
    nbins=BINS,
    title="Distribution des longueurs disciplinaires (distance totale)",
    labels={"distance": "Distance totale (longueur de trajectoire)"},
    marginal="rug",
    opacity=0.9,
)
fig_dist.update_layout(bargap=0.05)
fig_dist.show()

# ===============================
# 8) Sauvegarde HTML
# ===============================
fig_scatter_pub.write_html("scatter_publications_vs_speed_per_publication.html")
fig_speed.write_html("histogram_avg_speed_per_publication.html")
fig_dist.write_html("histogram_distance_per_publication.html")

print("✅ Graphiques sauvegardés :")
print("  - scatter_publications_vs_speed_per_publication.html")
print("  - histogram_avg_speed_per_publication.html")
print("  - histogram_distance_per_publication.html")
```

```python
# -*- coding: utf-8 -*-
import pandas as pd
import numpy as np
import plotly.express as px
from collections import defaultdict
import random

# ===============================
# 0) Paramètres
# ===============================
PATH_COORDS = "../data/coordinates.csv"             # colonnes: disc, dim1, dim2, ...
PATH_PUBS   = "../data/author_publications_dir/X.parquet"  # colonnes: author, discipline, year
MIN_PUBS_FILTER = 50       # seuil d'éligibilité des auteurs (filtre initial)
SAMPLE_MAX = 5000          # échantillon max d'auteurs
BINS = 50                  # nb de bacs pour les histogrammes
SEED = 42                  # reproductibilité

random.seed(SEED)
np.random.seed(SEED)

# ===============================
# 1) Chargement des coordonnées t-SNE/UMAP
# ===============================
df_coords = pd.read_csv(PATH_COORDS)
coord_cols = [c for c in df_coords.columns if c != "disc"]
disc_to_coord = {
    row["disc"]: row[coord_cols].to_numpy(dtype=np.float32)
    for _, row in df_coords.iterrows()
}

# ===============================
# 2) Chargement des publications
# ===============================
df_pub = pd.read_parquet(PATH_PUBS)
# Colonnes attendues : ['author', 'discipline', 'year']

# ===============================
# 3) Trajectoire "publication par publication"
#    + Distances normalisées (par publication & par pas)
# ===============================
def compute_distances_per_publication(author_name: str):
    """
    Calcule la trajectoire en ajoutant les publications une par une (ordre chronologique).
    Retourne:
      - total_pubs
      - total_distance : somme des distances entre barycentres successifs
      - avg_per_pub    : total_distance / total_pubs
      - avg_per_step   : total_distance / max(total_pubs - 1, 1)   (par segment)
    """
    sub = df_pub[df_pub["author"].str.lower() == author_name.lower()]
    if sub.empty:
        return 0, 0.0, 0.0, 0.0

    total_pubs = len(sub)
    if total_pubs <= 20:  # petit filtre anti-bruit
        return total_pubs, 0.0, 0.0, 0.0

    # tri chronologique (si égalité intra-année, l'ordre reste celui du DataFrame)
    sub_sorted = sub.sort_values(by=["year"]).reset_index(drop=True)

    trajectory = []
    aggregated = defaultdict(int)

    for _, row in sub_sorted.iterrows():
        disc = row["discipline"]
        aggregated[disc] += 1  # on ajoute UNE publi

        total = sum(aggregated.values())
        if total == 0:
            continue

        weighted = []
        for d, w in aggregated.items():
            if d in disc_to_coord:
                weighted.append(disc_to_coord[d] * w)
        if not weighted:
            continue

        coords = np.sum(weighted, axis=0) / total
        trajectory.append(coords)

    if len(trajectory) < 2:
        return total_pubs, 0.0, 0.0, 0.0

    trajectory = np.vstack(trajectory)
    diffs = trajectory[1:] - trajectory[:-1]
    total_distance = float(np.sum(np.linalg.norm(diffs, axis=1)))

    # Normalisations
    avg_per_pub  = total_distance / total_pubs if total_pubs > 0 else 0.0
    denom_steps  = max(total_pubs - 1, 1)
    avg_per_step = total_distance / denom_steps

    return total_pubs, total_distance, avg_per_pub, avg_per_step

# ===============================
# 4) Auteurs éligibles & échantillon
# ===============================
pub_counts = df_pub.groupby("author").size()
eligible_authors = pub_counts[pub_counts > MIN_PUBS_FILTER].index.to_numpy()

if len(eligible_authors) > SAMPLE_MAX:
    sample_authors = np.random.choice(eligible_authors, size=SAMPLE_MAX, replace=False)
else:
    sample_authors = eligible_authors

# ===============================
# 5) Calculs (granularité "par publication")
# ===============================
records = []
for author in sample_authors:
    total_pubs, total_dist, avg_per_pub, avg_per_step = compute_distances_per_publication(author)
    if total_pubs > 20:
        records.append({
            "author": author,
            "publications": int(total_pubs),
            "distance": float(total_dist),
            "avg_per_pub": float(avg_per_pub),
            "avg_per_step": float(avg_per_step),
        })

df_analysis_pub = pd.DataFrame(records)
if df_analysis_pub.empty:
    raise RuntimeError("df_analysis_pub est vide. Vérifie chemins/données/filtres.")

# ===============================
# 6) Corrélations
# ===============================
corr_pub_pub   = df_analysis_pub[["publications", "avg_per_pub"]].corr().iloc[0, 1]
corr_pub_step  = df_analysis_pub[["publications", "avg_per_step"]].corr().iloc[0, 1]
print(f"Corrélation (pubs vs distance moyenne PAR PUBLICATION, auteurs > {MIN_PUBS_FILTER} pubs) : {corr_pub_pub:.3f}")
print(f"Corrélation (pubs vs distance moyenne PAR PAS, auteurs > {MIN_PUBS_FILTER} pubs)         : {corr_pub_step:.3f}")

# ===============================
# 7) Scatter plots
# ===============================
fig_scatter_avgpub = px.scatter(
    df_analysis_pub,
    x="publications",
    y="avg_per_pub",
    log_x=True,
    hover_name="author",
    title="Distance moyenne par publication — granularité par publication",
    labels={
        "publications": "Nombre de publications",
        "avg_per_pub": "Distance moyenne / publication",
    },
)
fig_scatter_avgpub.show()

fig_scatter_avgstep = px.scatter(
    df_analysis_pub,
    x="publications",
    y="avg_per_step",
    log_x=False,
    hover_name="author",
    title="Distance moyenne par pas — granularité par publication",
    labels={
        "publications": "Nombre de publications",
        "avg_per_step": "Distance moyenne / pas (segments)",
    },
)
fig_scatter_avgstep.show()

# ===============================
# 8) Distributions (histogrammes)
# ===============================
fig_hist_avgpub = px.histogram(
    df_analysis_pub,
    x="avg_per_pub",
    nbins=BINS,
    title="Distribution — distance moyenne par publication",
    labels={"avg_per_pub": "Distance moyenne / publication"},
    marginal="rug",
    opacity=0.9,
)
fig_hist_avgpub.update_layout(bargap=0.05)
fig_hist_avgpub.show()

fig_hist_avgstep = px.histogram(
    df_analysis_pub,
    x="avg_per_step",
    nbins=BINS,
    title="Distribution — distance moyenne par pas",
    labels={"avg_per_step": "Distance moyenne / pas (segments)"},
    marginal="rug",
    opacity=0.9,
)
fig_hist_avgstep.update_layout(bargap=0.05)
fig_hist_avgstep.show()

fig_hist_dist = px.histogram(
    df_analysis_pub,
    x="distance",
    nbins=BINS,
    title="Distribution — longueur disciplinaire totale",
    labels={"distance": "Distance totale (longueur de trajectoire)"},
    marginal="rug",
    opacity=0.9,
)
fig_hist_dist.update_layout(bargap=0.05)
fig_hist_dist.show()

# ===============================
# 9) Sauvegardes HTML
# ===============================
fig_scatter_avgpub.write_html("scatter_publications_vs_avg_distance_per_publication.html")
fig_scatter_avgstep.write_html("scatter_publications_vs_avg_distance_per_step.html")
fig_hist_avgpub.write_html("histogram_avg_distance_per_publication.html")
fig_hist_avgstep.write_html("histogram_avg_distance_per_step(unlog).html")
fig_hist_dist.write_html("histogram_distance_total.html")

print("✅ Graphiques sauvegardés :")
print("  - scatter_publications_vs_avg_distance_per_publication.html")
print("  - scatter_publications_vs_avg_distance_per_step.html")
print("  - histogram_avg_distance_per_publication.html")
print("  - histogram_avg_distance_per_step.html")
print("  - histogram_distance_total.html")
```
